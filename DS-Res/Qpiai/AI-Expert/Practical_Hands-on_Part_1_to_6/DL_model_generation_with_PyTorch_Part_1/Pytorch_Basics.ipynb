## Broadcasting

When operating on two arrays, pytorch compares their shapes element-wise. It starts with the trailing (i.e. rightmost) dimensions and works its way left. Two dimensions are compatible when

1. they are equal.
2. one of them is a scaler

```python
a, b = torch.arange(5), torch.tensor(5)
a, b
```

here the scaler object `b` is broadcasted to the shape of a and the element wise addition takes place

```python
a + b
```

#### to replicate the same semantics in pytorch

```python
b.expand_as(a)
```

```python
assert torch.all(a + b.expand_as(a) == b + a)
```

## Matrix Multiplication

lets have a general idea of matrix multiplication implemented in plain pytorch and see how pytorch helps in speeding up processes

for a more visual introduction to matrix multiplication you can visit the following site

[http://matrixmultiplication.xyz](http://matrixmultiplication.xyz)

here we will implement a matrix multiplication operation in pytorch



```python
def matmul(a,b):
    ar,ac = a.shape # n_rows * n_cols
    br,bc = b.shape
    assert ac==br
    c = torch.zeros(ar, bc)
    for i in range(ar):
        for j in range(bc):
            for k in range(ac): # or br
                c[i,j] += a[i,k] * b[k,j]
    return c
```

```python
matmul(a, b)
```

```python
torch.matmul(a, b)
```

#### Here we can clearly see how efficient pytorch speeds up the code than the naive implementation of matrix multiplication

which is 10 times faster and the difference is even outstanding when the matrix is even larger

```python
%time matmul(a, b)
```

```python
%time torch.matmul(a, b)
```

```python
a, b = torch.randn((100, 100)), torch.randn((100, 100))
a, b
```

```python
%time matmul(a, b)
```

```python
%time torch.matmul(a, b)
```

## Pytorch views

PyTorch allows a tensor to be a View of an existing tensor. View tensor shares the same underlying data with its base tensor. Supporting View avoids explicit data copy, thus allows us to do fast and memory efficient reshaping, slicing and element-wise operations.

```python
a = torch.arange(25)
a
```

```python
a.view(-1, 25).shape
```

```python
a.view(5, 5)
```
